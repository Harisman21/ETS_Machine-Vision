```python
import cv2
import pytesseract

# Tentukan path ke tesseract
pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'  # Pastikan path benar

# Load gambar yang diunggah
image = cv2.imread('gambar2.jpeg')  # Gambar yang Anda unggah
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Terapkan threshold untuk meningkatkan kontras
_, thresh = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY)

# Gunakan deteksi tepi dengan Canny untuk menemukan kontur
edges = cv2.Canny(thresh, 100, 200)

# Temukan kontur dalam gambar
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Inisialisasi flag untuk menandakan apakah plat nomor terdeteksi
detected_plate = False

# Iterasi melalui kontur yang ditemukan
for contour in contours:
    # Dapatkan persegi panjang pembatas dari setiap kontur
    x, y, w, h = cv2.boundingRect(contour)

    # Tentukan rasio aspek untuk mendeteksi bentuk persegi panjang yang panjang (plat nomor)
    aspect_ratio = float(w) / h
    
    # Jika rasio aspek sesuai dengan plat nomor, gambar kotak di sekitar plat nomor
    if aspect_ratio > 2 and aspect_ratio < 6:  # Sesuaikan nilai ini dengan rasio plat nomor di gambar
        cv2.rectangle(image, (x, y), (x + w, y + h), (0, 255, 0), 2)
        
        # Crop area plat nomor untuk OCR
        plate_image = image[y:y+h, x:x+w]

        # Gunakan tesseract untuk membaca teks pada plat nomor
        plate_text = pytesseract.image_to_string(plate_image, config='--psm 8')

        if plate_text.strip():  # Jika OCR berhasil membaca plat nomor
            print(f"Plat nomor terdeteksi: {plate_text.strip()}")
        else:
            print("Plat nomor tidak terdeteksi menggunakan OCR.")
        
        detected_plate = True

# Jika tidak ada plat nomor terdeteksi
if not detected_plate:
    print("Tidak ada plat nomor yang terdeteksi.")

# Menampilkan gambar dengan deteksi
cv2.imshow('Detected Plate', image)

# Tunggu hingga tombol ditekan
cv2.waitKey(0)
cv2.destroyAllWindows()
```
